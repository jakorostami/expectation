```python
import numpy as np
import matplotlib.pyplot as plt
from expectation.parametric.ztest import SaviZtestStat
```

```python
# Example configuration
# observed z-statistic
z = 2.0               
# test-defining parameter (phiS, g, etc.)
parameter = 0.3        
# choose from: "grow", "egauss", "mom", "imom", "ecauchy"
etype = "grow"         
# options: "two_sided", "greater", "less"
alternative = "two_sided"  
ztest = SaviZtestStat(
    z=z,
    parameter=parameter,
    n1=50,
    etype=etype,
    alternative=alternative
)

e_value = ztest.compute_e_value()
print(f"E-value for {etype} (z={z}, n1=50, parameter={parameter}): {e_value}")
```
